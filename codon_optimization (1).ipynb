# Codon Optimization Project
DNA/protein sequence optimizer for E. coli, B. subtilis, P. aeruginosa, and M. tuberculosis.

## Stage 1 — Sequence validation, translation, GC content

```python
from Bio.Seq import Seq

def validate_dna(sequence: str) -> str:
    seq = sequence.strip().upper()
    valid_bases = set("ATGC")
    if not set(seq).issubset(valid_bases):
        raise ValueError(f"Invalid characters found: {set(seq) - valid_bases}")
    if len(seq) % 3 != 0:
        raise ValueError(f"Sequence length ({len(seq)}) is not divisible by 3")
    return seq


def translate_dna(sequence: str) -> str:
    seq = validate_dna(sequence)
    return str(Seq(seq).translate(to_stop=True))


def split_into_codons(sequence: str) -> list[str]:
    seq = validate_dna(sequence)
    return [seq[i:i+3] for i in range(0, len(seq), 3)]


def gc_content(sequence: str) -> float:
    seq = sequence.strip().upper()
    return (seq.count("G") + seq.count("C")) / len(seq) * 100
```

## Stage 3 — Naive optimizer
Picks the highest-frequency codon per amino acid for the chosen organism.

```python
def optimize_sequence(dna_sequence: str, codon_table: dict) -> str:
    protein = translate_dna(dna_sequence)
    optimized_codons = [best_codon(aa, codon_table) for aa in protein]
    optimized_codons.append(best_codon("*", codon_table))
    return "".join(optimized_codons)
```

## Stage 4 — Verification
Confirms the optimized DNA still encodes the identical protein.

```python
def verify_translation(original_dna: str, optimized_dna: str) -> bool:
    original_protein = translate_dna(original_dna)
    optimized_protein = translate_dna(optimized_dna)
    if original_protein != optimized_protein:
        raise ValueError(
            f"Protein mismatch!\nOriginal:  {original_protein}\nOptimized: {optimized_protein}"
        )
    return True
```

## Stage 5 — Codon Adaptation Index (CAI)

```python
import math

def _build_relative_adaptiveness(codon_table: dict) -> dict:
    weights = {}
    for aa, codon_dict in codon_table.items():
        if aa == "*":
            continue
        max_freq = max(codon_dict.values())
        for codon, freq in codon_dict.items():
            weights[codon] = freq / max_freq
    return weights


def calculate_cai(dna_sequence: str, codon_table: dict) -> float:
    weights = _build_relative_adaptiveness(codon_table)
    codons = split_into_codons(dna_sequence)
    w_values = [weights[c] for c in codons if c in weights]
    if not w_values:
        raise ValueError("No valid (non-stop) codons found to calculate CAI")
    log_sum = sum(math.log(w) for w in w_values)
    return math.exp(log_sum / len(w_values))
```

## Stage 6 — Comparison report

```python
def compare_sequences(original_dna: str, optimized_dna: str, codon_table: dict) -> dict:
    verify_translation(original_dna, optimized_dna)

    original_codons = split_into_codons(original_dna)
    optimized_codons = split_into_codons(optimized_dna)
    changed_positions = [
        i for i, (orig, opt) in enumerate(zip(original_codons, optimized_codons))
        if orig != opt
    ]

    return {
        "original_dna": original_dna,
        "optimized_dna": optimized_dna,
        "protein": translate_dna(original_dna),
        "cai_original": calculate_cai(original_dna, codon_table),
        "cai_optimized": calculate_cai(optimized_dna, codon_table),
        "gc_original": gc_content(original_dna),
        "gc_optimized": gc_content(optimized_dna),
        "num_codons": len(original_codons),
        "num_codons_changed": len(changed_positions),
        "percent_codons_changed": len(changed_positions) / len(original_codons) * 100,
        "changed_positions": changed_positions,
    }


def print_report(report: dict) -> None:
    print("=" * 50)
    print("CODON OPTIMIZATION REPORT")
    print("=" * 50)
    print(f"Protein sequence:      {report['protein']}")
    print(f"Original DNA:          {report['original_dna']}")
    print(f"Optimized DNA:         {report['optimized_dna']}")
    print("-" * 50)
    print(f"CAI (original):        {report['cai_original']:.4f}")
    print(f"CAI (optimized):       {report['cai_optimized']:.4f}")
    print(f"GC% (original):        {report['gc_original']:.2f}%")
    print(f"GC% (optimized):       {report['gc_optimized']:.2f}%")
    print("-" * 50)
    print(f"Total codons:          {report['num_codons']}")
    print(f"Codons changed:        {report['num_codons_changed']} "
          f"({report['percent_codons_changed']:.1f}%)")
    print(f"Changed positions:     {report['changed_positions']}")
    print("=" * 50)
```

```python
ecoli_table = get_codon_table("ecoli")

original_seq = "ATGGCTAAATAA"
optimized_seq = optimize_sequence(original_seq, ecoli_table)

print_report(compare_sequences(original_seq, optimized_seq, ecoli_table))
```

## Stage 7 — LP/MILP optimizer
Maximizes codon preference while constraining GC content to a target range (scipy MILP).

```python
import numpy as np
from scipy.optimize import milp, LinearConstraint, Bounds


def optimize_sequence_lp(dna_sequence: str, codon_table: dict,
                          gc_min: float = 0.0, gc_max: float = 100.0) -> str:
    protein = translate_dna(dna_sequence) + "*"
    position_codons = [list(codon_table[aa].keys()) for aa in protein]

    var_codon, var_position, costs = [], [], []
    for i, codons in enumerate(position_codons):
        for c in codons:
            var_codon.append(c)
            var_position.append(i)
            costs.append(-codon_table[protein[i]][c])

    n_vars = len(var_codon)
    n_positions = len(protein)
    costs = np.array(costs)

    A_select = np.zeros((n_positions, n_vars))
    for var_idx, pos in enumerate(var_position):
        A_select[pos, var_idx] = 1
    select_constraint = LinearConstraint(A_select, lb=1, ub=1)

    gc_counts = np.array([c.count("G") + c.count("C") for c in var_codon])
    total_bases = 3 * n_positions
    gc_constraint = LinearConstraint(
        gc_counts, lb=(gc_min / 100.0) * total_bases, ub=(gc_max / 100.0) * total_bases
    )

    bounds = Bounds(lb=0, ub=1)
    integrality = np.ones(n_vars)

    result = milp(costs, constraints=[select_constraint, gc_constraint],
                  bounds=bounds, integrality=integrality)
    if not result.success:
        raise ValueError(f"LP optimization infeasible: {result.message}")

    chosen = np.round(result.x).astype(int)
    optimized_codons = [None] * n_positions
    for var_idx, is_chosen in enumerate(chosen):
        if is_chosen:
            optimized_codons[var_position[var_idx]] = var_codon[var_idx]

    return "".join(optimized_codons)
```

```python
gc_test_protein = "MFLKRG"
gc_test_seq = "ATG" + "TTT" + "CTA" + "AAA" + "AGA" + "GGA" + "TAA"
assert translate_dna(gc_test_seq) == gc_test_protein

greedy_result = optimize_sequence(gc_test_seq, ecoli_table)
lp_result = optimize_sequence_lp(gc_test_seq, ecoli_table, gc_min=55, gc_max=70)

print("Original DNA:              ", gc_test_seq, f"(GC={gc_content(gc_test_seq):.1f}%)")
print("Greedy-optimized DNA:      ", greedy_result, f"(GC={gc_content(greedy_result):.1f}%)")
print("LP-optimized DNA (55-70%): ", lp_result, f"(GC={gc_content(lp_result):.1f}%)")
print()
print_report(compare_sequences(gc_test_seq, lp_result, ecoli_table))
```
